# 5.1 Features of Good Database Design

## 1. Simplicity
- The database structure should be easy to understand and manage.  
- Tables should represent clear and specific entities.  

## 2. Minimum Data Redundancy
- Duplicate data should be minimized.  
- Reduces storage waste and prevents inconsistencies.  

## 3. Data Integrity
- Data stored in the database should remain accurate and consistent.  
- Constraints such as **primary keys** and **foreign keys** help maintain integrity.  

## 4. Avoidance of Anomalies
A good design prevents:  
- **Insertion anomaly** – difficulty adding new data.  
- **Deletion anomaly** – accidental loss of important data.  
- **Update anomaly** – inconsistent updates in multiple places.  

## 5. Proper Use of Keys
- Each table should have a **primary key** for unique identification.  
- **Foreign keys** should properly establish relationships between tables.  

## 6. Normalization
- Tables should be normalized to reduce redundancy and dependency problems.  
- Common normal forms include **1NF, 2NF, and 3NF**.  

## 7. Data Consistency
- Same data should produce the same result throughout the database.  
- Ensures reliable information retrieval.  

## 8. Flexibility and Scalability
- The design should support future modifications and expansion.  
- New tables or attributes can be added easily.  

## 9. Efficient Query Performance
- Data should be organized for fast retrieval and updates.  
- Proper **indexing** improves performance.  

## 10. Security and Access Control
- Sensitive data should be protected.  
- Different users can have different access permissions.  


# 5.2 Keys in Relational Database

Keys are attributes used to identify records uniquely in a table and establish relationships between tables.

---

# 1. Super Key

A **Super Key** is a set of one or more attributes that can uniquely identify a tuple (row) in a table.

- It may contain extra attributes.
- Every candidate key is a super key, but every super key is not a candidate key.

## Example

### Student Table

| StudentID | Name | Email |
|---|---|---|
| 101 | Ram | ram@gmail.com |
| 102 | Sita | sita@gmail.com |

Possible Super Keys:
- {StudentID}
- {Email}
- {StudentID, Name}
- {StudentID, Email}

Because each combination can uniquely identify a student.

---

# 2. Candidate Key

A **Candidate Key** is a minimal super key.

- It uniquely identifies each row.
- It does not contain unnecessary attributes.
- A table can have multiple candidate keys.

## Example

From the Student table:

Candidate Keys:
- {StudentID}
- {Email}

Both uniquely identify records and are minimal.

But:
- {StudentID, Name} is not a candidate key because `Name` is unnecessary.

---

# 3. Primary Key

A **Primary Key** is the candidate key selected to uniquely identify tuples in a table.

## Features of Primary Key
- Must contain unique values.
- Cannot contain NULL values.
- Only one primary key is allowed in a table.

## Example

### Student Table

| StudentID (PK) | Name | Email |
|---|---|---|
| 101 | Ram | ram@gmail.com |
| 102 | Sita | sita@gmail.com |

Here:
- `StudentID` is chosen as the Primary Key.

---

# Difference Between Super Key, Candidate Key, and Primary Key

| Basis | Super Key | Candidate Key | Primary Key |
|---|---|---|---|
| Definition | Uniquely identifies rows | Minimal super key | Selected candidate key |
| Redundant Attributes | May contain | No | No |
| Number Allowed | Many | Multiple | Only one |
| NULL Values | Possible | Usually not preferred | Not allowed |

---

# Conclusion

- **Super Key** uniquely identifies rows but may include extra attributes.
- **Candidate Key** is the minimal form of a super key.
- **Primary Key** is the selected candidate key used for unique identification in the table.

# 5.3 Functional Dependencies

A **Functional Dependency (FD)** is a relationship between attributes in a relation where one attribute uniquely determines another attribute.

It is represented as:

A → B

This means:
- Attribute **A** determines attribute **B**.
- If two rows have the same value of A, they must have the same value of B.

---

# Definition

In a relation R, an attribute **Y** is functionally dependent on attribute **X** if each value of X is associated with exactly one value of Y.

Symbolically:

X → Y

Where:
- X = Determinant
- Y = Dependent attribute

---

# Example of Functional Dependency

## Student Table

| StudentID | Name | Department |
|---|---|---|
| 101 | Ram | CS |
| 102 | Sita | BBA |
| 103 | Hari | CS |

### Functional Dependencies
- StudentID → Name
- StudentID → Department

Explanation:
- Each StudentID uniquely determines the student's name and department.

---

# Types of Functional Dependencies

## 1. Full Functional Dependency

An attribute is fully functionally dependent on a key if it depends on the whole key and not on part of it.

### Example

| StudentID | CourseID | Grade |
|---|---|---|
| 101 | C101 | A |
| 102 | C102 | B |

Functional Dependency:

(StudentID, CourseID) → Grade

Here:
- Grade depends on both StudentID and CourseID together.

---

## 2. Partial Functional Dependency

A dependency where a non-key attribute depends on only part of a composite key.

### Example

| StudentID | CourseID | StudentName |
|---|---|---|
| 101 | C101 | Ram |

Functional Dependency:

StudentID → StudentName

Here:
- StudentName depends only on StudentID, not the whole composite key.

---

## 3. Transitive Dependency

When one non-key attribute depends on another non-key attribute.

### Example

| StudentID | DeptID | DeptName |
|---|---|---|
| 101 | D1 | Computer Science |

Functional Dependencies:
- StudentID → DeptID
- DeptID → DeptName

Therefore:
- StudentID → DeptName (Transitive Dependency)

---


## 4. Trivial Functional Dependency

A functional dependency is called **trivial** if the right-hand side (Y) is already a subset of the left-hand side (X).

### Rule

X → Y is trivial if:

Y ⊆ X

### Example

### Student Table

| StudentID | Name | Course |
|------------|------|--------|
| 101 | Ram | CS |
| 102 | Sita | IT |

### Functional Dependency

(StudentID, Name) → Name

This is **trivial** because **Name** already exists on the left-hand side.

Another example:

(StudentID, Course) → StudentID

This is also trivial.

### Why is it called trivial?

Because it is always true automatically.

---

## 5. Non-Trivial Functional Dependency

A functional dependency is called **non-trivial** if the right-hand side is **not** a subset of the left-hand side.

### Rule

X → Y is non-trivial if:

Y ⊄ X

### Example

### Student Table

| StudentID | Name | Course |
|------------|------|--------|
| 101 | Ram | CS |
| 102 | Sita | IT |

### Functional Dependency

StudentID → Name

This is **non-trivial** because **Name** is not part of the left-hand side.

Another example:

StudentID → Course

This is also non-trivial.

## Meaning

Knowing the StudentID helps determine the Name or Course.

---

## Difference Between Trivial and Non-Trivial Functional Dependency

| Trivial FD | Non-Trivial FD |
|-------------|----------------|
| RHS is part of LHS | RHS is not part of LHS |
| Always true | Gives meaningful relationship |
| Example: AB → A | Example: A → B |

---

## Completely Non-Trivial Functional Dependency

A dependency is called **completely non-trivial** if no attribute on the right-hand side appears on the left-hand side.

## Example

StudentID → Name

This is completely non-trivial because there is no common attribute on both sides.

But:

(StudentID, Name) → (Name, Course)

is only non-trivial, not completely non-trivial, because **Name** appears on both sides.

## Importance of Functional Dependency

- Helps in normalization.
- Reduces data redundancy.
- Prevents update anomalies.
- Maintains data consistency and integrity.
- Helps identify candidate keys.

---

## Advantages of Functional Dependency

- Improves database design.
- Ensures accurate data organization.
- Makes normalization easier.
- Reduces duplicate data.

---

## Conclusion

Functional Dependency describes the relationship between attributes in a relational database. It is an important concept used in normalization to reduce redundancy and maintain consistency in database systems.

# 5.4 Anomalies in Database

Database anomalies are problems that occur in a database due to poor design and data redundancy. These anomalies can lead to inconsistency and loss of data.

Anomalies usually occur when a table is not properly normalized.

---

# Types of Anomalies

1. Insertion Anomaly  
2. Update Anomaly  
3. Deletion Anomaly  

---

# 1. Insertion Anomaly

An insertion anomaly occurs when certain data cannot be inserted into the database without the presence of other data.

## Example

| StudentID | StudentName | Course |
|---|---|---|
| 101 | Ram | DBMS |
| 102 | Sita | OS |

Suppose a new course "AI" is introduced, but no student has enrolled yet.

Problem:
- The course information cannot be inserted because StudentID is required.

### Result
- Difficulty inserting data independently.

---

# 2. Update Anomaly

An update anomaly occurs when the same data is repeated in multiple rows and updating one row causes inconsistency.

## Example

| StudentID | StudentName | Department |
|---|---|---|
| 101 | Ram | CS |
| 102 | Sita | CS |

If the department name "CS" changes to "Computer Science", all rows must be updated.

Problem:
- If one row is missed, inconsistent data appears.

### Result
- Data inconsistency.

---

# 3. Deletion Anomaly

A deletion anomaly occurs when deleting one record also deletes important information unintentionally.

## Example

| StudentID | StudentName | Course |
|---|---|---|
| 101 | Ram | DBMS |

If student Ram leaves and the row is deleted:

Problem:
- Course information "DBMS" is also lost.

### Result
- Unintentional loss of valuable data.

---

# Causes of Anomalies

- Data redundancy
- Poor database design
- Lack of normalization

---

# How to Prevent Anomalies

- Apply normalization techniques.
- Divide large tables into smaller related tables.
- Use primary keys and foreign keys properly.

---

# Advantages of Removing Anomalies

- Improves data consistency
- Reduces redundancy
- Prevents data loss
- Makes database maintenance easier

---

# Conclusion

Anomalies are problems caused by improper database design. Proper normalization helps eliminate insertion, update, and deletion anomalies, resulting in a more efficient and reliable database system.

# 5.5 Decomposition Using Functional Dependencies

## Introduction

Decomposition is the process of dividing a relation (table) into two or more smaller relations to reduce redundancy and remove anomalies.

Functional dependencies are used to determine how a table should be decomposed during normalization.

---

# Objectives of Decomposition

- Reduce data redundancy
- Eliminate anomalies
- Improve data consistency
- Achieve normalization
- Maintain data integrity

---

# Functional Dependency

A functional dependency is represented as:

X → Y

Meaning:
- Attribute X determines attribute Y.

Example:

StudentID → StudentName

Here:
- StudentID uniquely determines StudentName.

---

# Example of Decomposition

## Original Relation

### Student_Course Table

| StudentID | StudentName | CourseID | CourseName |
|---|---|---|---|
| 101 | Ram | C101 | DBMS |
| 102 | Sita | C102 | OS |

### Functional Dependencies

- StudentID → StudentName
- CourseID → CourseName

Problem:
- Student and course information are repeated.
- Causes redundancy and anomalies.

---

# Decomposition Process

The original table can be decomposed into two smaller tables.

---

## 1. Student Table

| StudentID | StudentName |
|---|---|
| 101 | Ram |
| 102 | Sita |

Functional Dependency:
- StudentID → StudentName

---

## 2. Course Table

| CourseID | CourseName |
|---|---|
| C101 | DBMS |
| C102 | OS |

Functional Dependency:
- CourseID → CourseName

---

# Benefits of Decomposition

- Removes redundancy
- Eliminates insertion, update, and deletion anomalies
- Improves consistency
- Simplifies maintenance

---

# Properties of Good Decomposition

## 1. Lossless Join Decomposition

After decomposition, the original table can be reconstructed without losing information.

### Example

If:
- R → R1 and R2

Then joining R1 and R2 should reproduce the original relation.

### Advantage
- No data loss occurs.

---

## 2. Dependency Preservation

Functional dependencies should still be maintained after decomposition.

### Advantage
- Constraints can be enforced easily without joining tables.

---

# Example of Lossless Decomposition

## Original Relation

| StudentID | StudentName | DeptID | DeptName |
|---|---|---|---|
| 101 | Ram | D1 | CS |

Functional Dependencies:
- StudentID → StudentName
- DeptID → DeptName

---

## Decomposed Relations

### Student Table

| StudentID | StudentName | DeptID |
|---|---|---|
| 101 | Ram | D1 |

### Department Table

| DeptID | DeptName |
|---|---|
| D1 | CS |

The original relation can be reconstructed by joining the tables using DeptID.

---

# Advantages of Decomposition

- Better database organization
- Reduced duplication
- Easier updates
- Improved consistency
- Efficient normalization

---

# Disadvantages of Excessive Decomposition

- Too many tables may increase complexity.
- More joins may reduce query performance.

---

# Conclusion

Decomposition using functional dependencies is an important process in database normalization. It divides large relations into smaller relations to reduce redundancy and anomalies while preserving data integrity and consistency.

# 5.6 Database Normalization Notes (1NF, 2NF, 3NF, BCNF)

## 📌 What is Normalization?
- **Definition:** Process of structuring a relational database to reduce redundancy and improve data integrity.  
- **Goal:** Organize data into smaller, related tables to avoid anomalies (insertion, update, deletion).  
- **Benefits:** Eliminates duplicate data, enforces consistency, improves scalability, and optimizes query performance.  

---

## 1️⃣ First Normal Form (1NF)
**Rule:**  
- All attributes must be **atomic** (no lists, sets, or repeating groups).  
- Each row must be unique.  
- one column -> same data type 

**Example (Unnormalized):**
| CustomerID | CustomerName | PurchasedProducts       |
|------------|--------------|-------------------------|
| 101        | John Doe     | Laptop, Mouse           |
| 102        | Jane Smith   | Tablet                  |

❌ Problem: Multi-valued attribute `PurchasedProducts`.  

✅ **1NF Table:**
| CustomerID | CustomerName | Product   |
|------------|--------------|-----------|
| 101        | John Doe     | Laptop    |
| 101        | John Doe     | Mouse     |
| 102        | Jane Smith   | Tablet    |

---

## 2️⃣ Second Normal Form (2NF)
**Rule:**  
- Must satisfy 1NF.  
- No **partial dependency** (non-key attributes must depend on the whole primary key).  

**Example (1NF Issue):**
| OrderID | CustomerID | CustomerName | Product |
|---------|------------|--------------|---------|
| 201     | 101        | John Doe     | Laptop  |
| 202     | 101        | John Doe     | Mouse   |

❌ Problem: `CustomerName` depends only on `CustomerID`, not the composite key `(OrderID, CustomerID)`.  

✅ **2NF Tables:**
**Customers**
| CustomerID | CustomerName |
|------------|--------------|
| 101        | John Doe     |
| 102        | Jane Smith   |

**Orders**
| OrderID | CustomerID | Product |
|---------|------------|---------|
| 201     | 101        | Laptop  |
| 202     | 101        | Mouse   |

---

## 3️⃣ Third Normal Form (3NF)
**Rule:**  
- Must satisfy 2NF.  
- No **transitive dependency** (non-key attributes should not depend on other non-key attributes).  

**Example (2NF Issue):**
| OrderID | CustomerID | Product | Supplier   |
|---------|------------|---------|------------|
| 201     | 101        | Laptop  | HP         |
| 202     | 101        | Mouse   | Logitech   |
           

❌ Problem: `Supplier` depends on `Product`, not directly on `OrderID`.  
    Product -> Non key attribute (shouldn't depend on this ) 
    

✅ **3NF Tables:**
**Products**
| ProductID | ProductName | SupplierID |
|-----------|-------------|------------|
| 301       | Laptop      | 401        |
| 302       | Mouse       | 402        |

**Suppliers**
| SupplierID | SupplierName |
|------------|--------------|
| 401        | HP           |
| 402        | Logitech     |

**Orders**
| OrderID | CustomerID | ProductID |
|---------|------------|-----------|
| 201     | 101        | 301       |
| 202     | 101        | 302       |

---

## 4️⃣ Boyce-Codd Normal Form (BCNF)
**Rule:**  
- Must satisfy 3NF.  
- For every dependency **X → Y**, X must be a **candidate key**.  

## 🧩 Example of a Table Not in BCNF

| **Student** | **Course**     | **Teacher** |
|--------------|----------------|--------------|
| Sok          | DB             | John         |
| Sao          | DB             | William      |
| Chan         | E-Commerce     | Todd         |
| Sok          | E-Commerce     | Todd         |
| Chan         | DB             | William      |

---

### 🔑 Key
`{Student, Course}`

### ⚙️ Functional Dependency
- `{Student, Course} → Teacher`
- `Teacher → Course`

###2nf satisfied as no partial dependency > not depending on part of candidate key , but full candidate key (student,course) .
###3nf satisfied as no transitive dependency . (not dependent on NP attribute) .
###Fails bcnf
                                                                                 

### ⚠️ Problem
###*Teacher* is **not a superkey** but determines *Course*.


## 🧠 Solution (BCNF Decomposition)

**Step:**  
Split into two tables:  
1. **Student–Course**  
2. **Course–Teacher**

---

### 🗂️ Decomposed Tables

#### Table 1: Student–Course
| **Student** | **Course**     |
|--------------|----------------|
| Sok          | DB             |
| Sao          | DB             |
| Chan         | E-Commerce     |
| Sok          | E-Commerce     |
| Chan         | DB             |

#### Table 2: Course–Teacher
| **Course**     | **Teacher** |
|----------------|--------------|
| DB             | John         |
| DB             | William      |
| E-Commerce     | Todd         |
