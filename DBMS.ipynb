# What is a Database?
A database is an organized collection of data that can be easily accessed, managed, and updated.

Data is stored in a structured format (tables, rows, columns) to ensure efficiency and consistency.

Example: A university database storing student records, courses, and grades.

## What is DBMS (Database Management System)?

*DBMS is software that manages databases.*

It provides tools for:

+ Data definition (creating tables, schemas).

+ Data manipulation (insert, update, delete, query).

+ Data security (user authentication, access control).

+ Concurrency control (multiple users accessing data simultaneously).

Backup & recovery (restore data after failures).

Examples: MySQL, Oracle, SQL Server, PostgreSQL.

## What is RDBMS (Relational Database Management System)?
RDBMS is a type of DBMS based on the relational model (introduced by E.F. Codd in 1970).

Data is stored in tables (relations) with rows (tuples) and columns (attributes).

Uses SQL (Structured Query Language) for queries.

Supports:

+ Normalization (reducing redundancy).

+ Constraints (primary key, foreign key, unique).

+ Transactions (ACID properties: Atomicity, Consistency, Isolation, Durability).

Examples: MySQL, PostgreSQL, Oracle, SQL Server.

## Tables (Relations)
A table is the basic structure in a relational database.

It organizes data into a grid-like format, similar to a spreadsheet.

Each table represents one type of entity (e.g., Students, Employees, Products).

## Columns (Attributes)
Columns define the properties or characteristics of the entity.

Each column has a name and a data type (e.g., integer, text, date).

Example in a Students table:

- Student_ID (integer)
- Name (text)
- Age (integer)
- Course (text)

👉 Columns = fields that describe what kind of data is stored.

## Rows (Tuples)
Rows are individual records in the table.

Each row contains actual data values for the attributes.

Example row in Students table:

101, "Sita", 20, "Computer Science"

👉 Rows = entries that represent one instance of the entity.

**Table (Relation):** Students  
**Columns (Attributes):** Student_ID, Name, Age, Course

# So when we say “data is stored in tables with rows and columns”, it means that the database organizes information in a structured way where:

+ Columns describe the type of data.

+ Rows hold the actual data values.

Together, they form a relation (a complete dataset about one entity).

They enable scalable applications (from small websites to global enterprises).

They are the backbone of modern IT systems (banking, healthcare, e-commerce, social media).

  # 1.2 Purpose of Database Systems

Databases serve several important purposes in modern computing:

- **Data Management**: Provide efficient methods for storing, organizing, and retrieving large volumes of data.  
- **Consistency**: Reduce redundancy and maintain data integrity so that information remains accurate across multiple applications.  
- **Security**: Offer controlled access to sensitive information through authentication (who you are) and authorization (what you can do).  
- **Concurrency**: Allow multiple users to access and update data simultaneously without conflicts, ensuring smooth operations in multi-user environments.  
- **Backup & Recovery**: Safeguard against data loss by enabling restoration after hardware failures, software crashes, or human errors.  

# 👁️ 1.3 View of Data

Databases provide different levels of abstraction to simplify how users interact with data. These levels hide unnecessary details from users and make the system easier to use and manage.

---

## 🔹 Physical Level
- **Definition**: The lowest level of abstraction, describing how data is physically stored in memory or on disk.  
- **Details**: Includes files, indexes, storage blocks, and pointers.  
- **Users**: Database administrators and system designers.  
- **Example**: Data stored on disk as binary files, with indexing structures like B-trees to speed up searches.  

👉 Think of it as the **“behind-the-scenes” storage mechanism**.

---

## 🔹 Logical Level
- **Definition**: The middle level of abstraction, describing *what data is stored* and *what relationships exist* between data.  
- **Details**: Defines tables, attributes, constraints, and relationships (primary keys, foreign keys).  
- **Users**: Application programmers and database designers.  
- **Example**: A `Students` table with attributes like `Student_ID`, `Name`, `Age`, `Course`.  

👉 This is the **blueprint of the database**—it shows structure but hides physical storage details.

---

## 🔹 View Level
- **Definition**: The highest level of abstraction, describing *how users interact with data*.  
- **Details**: Provides customized views for different users, often through queries, forms, or reports.  
- **Users**: End users (casual or sophisticated).  
- **Example**:  
  - A student sees only their grades and personal info.  
  - A professor sees all students enrolled in their course.  
  - An administrator sees the entire database.  

👉 This is the **user-friendly interface**—it hides both logical and physical complexity.

---

# 📊 Example: University Database

| Student_ID | Name | Age | Course             |
|------------|------|-----|--------------------|
| 101        | Sita | 20  | Computer Science   |
| 102        | Ram  | 22  | Information Tech   |
| 103        | Gita | 19  | Business Studies   |

- **Physical Level**: Data stored as files on disk with indexes for quick lookup.  
- **Logical Level**: Defined as a table `Students(Student_ID, Name, Age, Course)`.  
- **View Level**:  
  - Student sees: `Name, Age, Course`.  
  - Professor sees: `Student_ID, Name, Course`.  
  - Admin sees: all columns.  

---

# ✅ Why Views of Data Matter
- **Simplifies interaction**: Users don’t need to know how data is stored.  
- **Security**: Different users can be restricted to see only relevant data.  
- **Flexibility**: Changes at the physical level don’t affect how users interact with data.  
- **Abstraction**: Each level hides unnecessary details from the next.  

👉 **Practical Note**:  
When you log into your bank account online, you only see your transactions (**view level**). The bank’s system knows the logical structure (tables for accounts, customers, transactions). The physical level is hidden completely—it’s just files and indexes on servers.




# 🏛️ 1.4 Database and Application Architecture

Databases can be integrated into applications using different architectures:

- **Two-tier Architecture**:  
  - Direct connection between client and database server.  
  - Example: A desktop application sending SQL queries directly to the database.  

- **Three-tier Architecture**:  
  - Client ↔ Application Server ↔ Database Server.  
  - Example: Web applications where the browser communicates with a web server, which then interacts with the database.  

- **Distributed Databases**:  
  - Data is spread across multiple physical locations but appears as a single logical database to users.  
  - Useful for global organizations with multiple branches. 

- **Cloud Databases**: 
  - Hosted on cloud platforms (AWS, Azure, Google Cloud).

---

# 👥 1.5 Database Users and Administrators

Different types of people interact with databases:

- **End Users**:  
  - *Casual users*: Use predefined queries, forms, or reports (e.g., ATM customers).  
  - *Sophisticated users*: Write complex queries for analysis (e.g., data analysts).  

- **Application Programmers**:  
  - Develop software applications that interact with databases.  
  - Example: A developer writing code for an e-commerce website that stores customer orders.  

- **Database Administrators (DBAs)**:  
  - Manage the overall database environment.  
  - Responsibilities include:  
    - Ensuring security (user accounts, permissions).  
    - Performance tuning (optimizing queries, indexing).  
    - Backup and recovery planning.  
    - Monitoring and troubleshooting.  



# 1.6 History of Database Systems

Databases have evolved over decades to meet the growing needs of organizations and technology.

---

## 📂 1960s: File-Based Systems
- Data stored in flat files (text or binary).  
- No structured query language; data access was manual and application-specific.  
- Problems: Redundancy, inconsistency, poor security, and difficulty in managing large datasets.  
- Example: Payroll records stored in separate files for each department.

---

## 🗂️ 1970s: Hierarchical & Network Models
- **Hierarchical Model**: Data organized in a tree-like structure (parent-child relationships).  
  - Example: An organization database where a department is the parent and employees are children.  
- **Network Model**: Data represented as records connected by links (graph-like structure).  
  - Example: Airline reservation systems.  
- Advantages: Faster access for certain queries.  
- Limitations: Complex to design, rigid structure, difficult to modify.

---

## 📊 1980s: Relational Model (E.F. Codd)
- Introduced by **E.F. Codd** in 1970, became widely adopted in the 1980s.  
- Data stored in **tables (relations)** with rows (tuples) and columns (attributes).  
- **SQL (Structured Query Language)** became the standard for querying and managing data.  
- Advantages: Simplicity, flexibility, reduced redundancy, strong theoretical foundation.  
- Examples: Oracle, MySQL, PostgreSQL, SQL Server.

---

## 🎥 1990s: Object-Oriented Databases
- Designed to handle complex data types (images, audio, video, multimedia).  
- Integrated database concepts with object-oriented programming.  
- Example: Object-relational databases that combine relational tables with object features.  
- Use cases: CAD/CAM, multimedia applications, scientific research.

---

## 🌐 2000s Onwards: Modern Database Systems
- **NoSQL Databases**:  
  - Handle unstructured or semi-structured data.  
  - Types: Document-oriented (MongoDB), Column-oriented (Cassandra), Key-value stores (Redis).  
  - Use cases: Social media, big data, real-time analytics.  

- **Cloud Databases**:  
  - Hosted on cloud platforms (AWS, Azure, Google Cloud).  
  - Scalable, accessible globally, pay-as-you-go model.  
  - Example: Amazon RDS, Google Cloud Spanner.  

- **Big Data Systems**:  
  - Designed for massive datasets and distributed processing.  
  - Tools: Hadoop, Spark.  
  - Use cases: Data mining, machine learning, predictive analytics.  

- **Modern Trends**:  
  - AI-driven databases (self-tuning, automated query optimization).  
  - Real-time analytics for instant decision-making.  
  - Integration with IoT (Internet of Things) and edge computing.

---

👉 **Practical Note**:  
- **Relational databases** (MySQL, Oracle, SQL Server, PostgreSQL) remain dominant for structured data and enterprise applications.  
- **NoSQL databases** (MongoDB, Cassandra) are popular for unstructured, large-scale, and high-speed applications like social media, IoT, and big data.  
- **Cloud databases** are increasingly adopted due to scalability and cost efficiency.
